## 3 — Dataset quality and answerability audit

```python
# Step 3: Dataset quality and answerability audit

import pandas as pd
import json

# Convert nested values into hashable representations
def make_hashable(value):
    if isinstance(value, (list, tuple)):
        return tuple(make_hashable(v) for v in value)

    if isinstance(value, dict):
        return tuple(sorted(
            (k, make_hashable(v)) for k, v in value.items()
        ))

    if isinstance(value, set):
        return tuple(sorted(make_hashable(v) for v in value))

    if value is None:
        return None

    try:
        if pd.isna(value):
            return None
    except (TypeError, ValueError):
        pass

    return value


def audit_dataset(df, split_name):
    print(f"\n{'=' * 60}")
    print(f"{split_name.upper()} DATASET AUDIT")
    print(f"{'=' * 60}")

    print("Total records:", len(df))

    # Missing values
    print("\nMissing values:")
    display(df.isnull().sum().to_frame("Missing Count"))

    # Empty strings
    string_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns

    empty_counts = {
        col: df[col].astype("string").str.strip().eq("").sum()
        for col in string_columns
    }

    print("\nEmpty string values:")
    display(pd.Series(empty_counts, name="Empty Count").to_frame())

    # Exact duplicate records (supports list/dict columns)
    row_keys = df.apply(
        lambda row: tuple(make_hashable(v) for v in row),
        axis=1
    )

    print("\nExact duplicate records:", row_keys.duplicated().sum())

    # Duplicate question IDs
    if "QUESTION_ID" in df.columns:
        print(
            "Duplicate QUESTION_ID values:",
            df["QUESTION_ID"].duplicated().sum()
        )

    # Duplicate question text
    if "QUESTION_TEXT" in df.columns:
        print(
            "Duplicate QUESTION_TEXT values:",
            df["QUESTION_TEXT"].duplicated().sum()
        )

    # Answerability distribution
    if "ANSWERABLE" in df.columns:
        print("\nAnswerability distribution:")

        counts = df["ANSWERABLE"].value_counts(dropna=False)

        distribution = pd.DataFrame({
            "Count": counts,
            "Percentage": (counts / len(df) * 100).round(2)
        })

        display(distribution)


# Audit both datasets
audit_dataset(train_df, "Training")
audit_dataset(dev_df, "Development")

# Compare answerability distributions
if "ANSWERABLE" in train_df.columns and "ANSWERABLE" in dev_df.columns:

    comparison = pd.concat({
        "Training": train_df["ANSWERABLE"].value_counts(normalize=True) * 100,
        "Development": dev_df["ANSWERABLE"].value_counts(normalize=True) * 100
    }, axis=1).fillna(0).round(2)

    print("\nANSWERABILITY COMPARISON (%)")
    display(comparison)
```

```python
# Step 5: Verify whether TechQA offsets are 1-based

def verify_offset_convention(df, split_name):
    results = {
        "Exact at stored offset": 0,
        "Exact at offset - 1": 0,
        "Not found at either position": 0,
        "Unanswerable": 0,
        "Invalid offset": 0
    }

    for _, row in df.iterrows():
        if str(row["ANSWERABLE"]).strip().upper() != "Y":
            results["Unanswerable"] += 1
            continue

        doc_id = str(row["DOCUMENT"]).strip()
        text = documents.get(doc_id, "")
        answer = str(row["ANSWER"])

        try:
            offset = int(row["START_OFFSET"])
        except (ValueError, TypeError):
            results["Invalid offset"] += 1
            continue

        if text[offset:offset + len(answer)] == answer:
            results["Exact at stored offset"] += 1
        elif offset > 0 and text[offset-1:offset-1+len(answer)] == answer:
            results["Exact at offset - 1"] += 1
        else:
            results["Not found at either position"] += 1

    print(f"\n{split_name}")
    display(pd.Series(results, name="Count").to_frame())


verify_offset_convention(train_df, "Training")
verify_offset_convention(dev_df, "Development")
```

## 6 — Create corrected answer offsets

```python
# Step 6: Normalize answer offsets without modifying originals

def normalize_offsets(df, split_name):
    df = df.copy()

    df["CORRECTED_START_OFFSET"] = pd.NA
    df["OFFSET_STATUS"] = "Unresolved"

    for idx, row in df.iterrows():

        if str(row["ANSWERABLE"]).strip().upper() != "Y":
            df.at[idx, "OFFSET_STATUS"] = "Unanswerable"
            continue

        doc_id = str(row["DOCUMENT"]).strip()
        text = documents.get(doc_id, "")
        answer = str(row["ANSWER"])

        try:
            offset = int(row["START_OFFSET"])
        except (ValueError, TypeError):
            continue

        if text[offset:offset + len(answer)] == answer:
            corrected = offset
            status = "Exact stored offset"

        elif (
            offset > 0
            and text[offset - 1:offset - 1 + len(answer)] == answer
        ):
            corrected = offset - 1
            status = "Corrected by -1"

        else:
            continue

        df.at[idx, "CORRECTED_START_OFFSET"] = corrected
        df.at[idx, "OFFSET_STATUS"] = status

    print(f"\n{split_name} OFFSET SUMMARY")
    display(df["OFFSET_STATUS"].value_counts().to_frame("Count"))

    return df


train_df = normalize_offsets(train_df, "Training")
dev_df = normalize_offsets(dev_df, "Development")
```

# 7 — Investigate unresolved answer spans

```python

# STEP 7: Final investigation and safe resolution of answer spans

import pandas as pd
import numpy as np

def resolve_answer_spans(df, split_name):
    result = df.copy()

    statuses = []
    verified_offsets = []
    candidate_offsets = []

    for _, row in result.iterrows():

        doc_id = str(row["DOCUMENT"]).strip()
        document = documents.get(doc_id, "")

        answer_value = row["ANSWER"]
        answer = "" if pd.isna(answer_value) else str(answer_value)

        offset_value = row["START_OFFSET"]

        # Safely read the original offset
        try:
            stored_offset = int(offset_value)
        except (ValueError, TypeError, OverflowError):
            stored_offset = None

        answerable = str(row["ANSWERABLE"]).strip().upper()

        # Unanswerable examples do not require answer spans
        if answerable == "N":
            statuses.append("Unanswerable")
            verified_offsets.append(None)
            candidate_offsets.append([])
            continue

        # Missing document
        if not isinstance(document, str) or not document:
            statuses.append("Document missing")
            verified_offsets.append(None)
            candidate_offsets.append([])
            continue

        # Answerable record with empty answer
        if not answer.strip():
            statuses.append("Empty answer annotation")
            verified_offsets.append(None)
            candidate_offsets.append([])
            continue

        # Find every exact occurrence in the document
        occurrences = []
        start = 0

        while True:
            pos = document.find(answer, start)
            if pos == -1:
                break
            occurrences.append(pos)
            start = pos + 1

        # Original offset is correct
        if stored_offset is not None and stored_offset in occurrences:
            statuses.append("Exact stored offset")
            verified_offsets.append(stored_offset)
            candidate_offsets.append(occurrences)

        # One unique exact match: safe to correct
        elif len(occurrences) == 1:
            statuses.append("Corrected unique exact match")
            verified_offsets.append(occurrences[0])
            candidate_offsets.append(occurrences)

        # Multiple occurrences: cannot choose arbitrarily
        elif len(occurrences) > 1:
            statuses.append("Ambiguous multiple matches")
            verified_offsets.append(None)
            candidate_offsets.append(occurrences)

        # Answer does not occur in linked document
        else:
            statuses.append("Answer not found in document")
            verified_offsets.append(None)
            candidate_offsets.append([])

    # Keep original annotations unchanged
    result["STEP7_STATUS"] = statuses
    result["VERIFIED_START_OFFSET"] = pd.array(
        verified_offsets, dtype="Int64"
    )
    result["CANDIDATE_OFFSETS"] = candidate_offsets

    print(f"\n{'='*55}")
    print(f"{split_name} SUMMARY")
    print(f"{'='*55}")
    print(result["STEP7_STATUS"].value_counts(dropna=False))
    print("Total records:", len(result))

    exceptions = result[
        ~result["STEP7_STATUS"].isin([
            "Unanswerable",
            "Exact stored offset",
            "Corrected unique exact match"
        ])
    ]

    print("\nRecords requiring manual review:", len(exceptions))

    if not exceptions.empty:
        print(
            exceptions[
                ["QUESTION_ID", "ANSWERABLE", "DOCUMENT",
                 "START_OFFSET", "ANSWER", "STEP7_STATUS",
                 "CANDIDATE_OFFSETS"]
            ].to_string(index=False, max_colwidth=100)
        )

    return result


train_df = resolve_answer_spans(train_df, "Training")
dev_df = resolve_answer_spans(dev_df, "Development")
```

# 8: Create preprocessing-ready TechQA datasets.

```python

# STEP 8: Prepare preprocessing-ready TechQA datasets

from pathlib import Path
import pandas as pd

OUTPUT_DIR = Path("../../data/processed/TechQA")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


def prepare_dataset(df, split_name):

    processed = df.copy()

    # Identify records with verified answer spans
    processed["SPAN_VALID"] = (
        processed["STEP7_STATUS"].isin([
            "Exact stored offset",
            "Corrected unique exact match"
        ])
    )

    # Store the verified offset separately.
    # Original START_OFFSET remains unchanged.
    processed["PROCESSED_START_OFFSET"] = (
        processed["VERIFIED_START_OFFSET"]
    )

    # Save the complete dataset without silently dropping records
    output_file = OUTPUT_DIR / f"{split_name}_processed.parquet"
    processed.to_parquet(output_file, index=False)

    print(f"\n{split_name.upper()} DATASET")
    print("Total records:", len(processed))
    print("Answerable:", (processed["ANSWERABLE"] == "Y").sum())
    print("Unanswerable:", (processed["ANSWERABLE"] == "N").sum())
    print("Verified answer spans:", processed["SPAN_VALID"].sum())
    print("Unverified answer spans:",
          ((processed["ANSWERABLE"] == "Y") &
           (~processed["SPAN_VALID"])).sum())
    print("Saved:", output_file)

    return processed


train_processed = prepare_dataset(train_df, "train")
dev_processed = prepare_dataset(dev_df, "dev")

print("\nSTEP 8 COMPLETED")
```

# 10 — Investigate and document duplicate records and annotation conflicts

```python
# Step 10: Investigate duplicate records and annotation conflicts

def normalize_value(value):
    """Convert list-like values into comparable strings."""
    if isinstance(value, (list, tuple, dict, set)):
        return repr(value)
    return value


def investigate_duplicates(df, split_name):

    print(f"\n{'='*60}")
    print(f"{split_name} DUPLICATE ANALYSIS")
    print(f"{'='*60}")

    # Create a comparison copy
    check_df = df.copy()

    # Normalize list-valued columns
    for col in check_df.columns:
        check_df[col] = check_df[col].apply(normalize_value)

    # Duplicate question IDs
    print(
        "Duplicate question IDs:",
        check_df["QUESTION_ID"].duplicated().sum()
    )

    # Duplicate question-document pairs
    pair_cols = ["QUESTION_TEXT", "DOCUMENT"]

    duplicate_pairs = check_df.duplicated(
        subset=pair_cols,
        keep=False
    )

    print(
        "Records in duplicate question-document pairs:",
        duplicate_pairs.sum()
    )

    # Exact duplicate records
    exact_duplicates = check_df.duplicated(keep=False)

    print(
        "Records involved in exact duplicates:",
        exact_duplicates.sum()
    )

    # Annotation conflicts
    conflicts = (
        check_df.groupby(pair_cols, dropna=False)
        .agg(
            answer_variants=("ANSWER", "nunique"),
            answerability_variants=("ANSWERABLE", "nunique"),
            record_count=("QUESTION_ID", "size")
        )
        .reset_index()
    )

    conflicts = conflicts[
        (conflicts["answer_variants"] > 1) |
        (conflicts["answerability_variants"] > 1)
    ]

    print(
        "Conflicting question-document groups:",
        len(conflicts)
    )

    if not conflicts.empty:
        print("\nSample conflicts:")
        display(conflicts.head(10))
    else:
        print("\nNo annotation conflicts found.")

    return conflicts


train_conflicts = investigate_duplicates(train_df, "TRAINING")
dev_conflicts = investigate_duplicates(dev_df, "DEVELOPMENT")

print("\nSTEP 10 COMPLETED")
```

# 11: Check dataset and document leakage across splits.

```python
import pandas as pd
import re

def normalize_text(value):
    if pd.isna(value):
        return ""
    value = str(value).lower().strip()
    return re.sub(r"\s+", " ", value)


def check_split_leakage(train_df, dev_df):
    train = train_df.copy()
    dev = dev_df.copy()

    # Normalize fields for reliable comparison
    for df in (train, dev):
        df["NORMALIZED_QUESTION"] = df["QUESTION_TEXT"].apply(
            normalize_text
        )
        df["NORMALIZED_DOCUMENT_ID"] = df["DOCUMENT"].apply(
            normalize_text
        )

    # 1. Question ID overlap
    train_ids = set(train["QUESTION_ID"].dropna().astype(str))
    dev_ids = set(dev["QUESTION_ID"].dropna().astype(str))
    common_ids = train_ids & dev_ids

    # 2. Exact normalized question overlap
    train_questions = set(train["NORMALIZED_QUESTION"]) - {""}
    dev_questions = set(dev["NORMALIZED_QUESTION"]) - {""}
    common_questions = train_questions & dev_questions

    # 3. Question-document pair overlap
    train_pairs = set(
        zip(train["NORMALIZED_QUESTION"],
            train["NORMALIZED_DOCUMENT_ID"])
    )
    dev_pairs = set(
        zip(dev["NORMALIZED_QUESTION"],
            dev["NORMALIZED_DOCUMENT_ID"])
    )
    common_pairs = train_pairs & dev_pairs

    # 4. Document ID overlap
    train_documents = set(train["NORMALIZED_DOCUMENT_ID"]) - {"", "-"}
    dev_documents = set(dev["NORMALIZED_DOCUMENT_ID"]) - {"", "-"}
    common_documents = train_documents & dev_documents

    print("=" * 65)
    print("TECHQA TRAINING–DEVELOPMENT LEAKAGE ANALYSIS")
    print("=" * 65)

    print("\nQUESTION ID OVERLAP")
    print("Common question IDs:", len(common_ids))

    print("\nQUESTION TEXT OVERLAP")
    print("Common normalized questions:", len(common_questions))

    print("\nQUESTION-DOCUMENT PAIR OVERLAP")
    print("Common question-document pairs:", len(common_pairs))

    print("\nDOCUMENT OVERLAP")
    print("Common document IDs:", len(common_documents))

    # Show overlapping examples, if any
    if common_questions:
        print("\nOverlapping questions (first 10):")
        for question in list(common_questions)[:10]:
            print("-", question[:200])

    if common_documents:
        print("\nShared document IDs (first 10):")
        for doc_id in list(common_documents)[:10]:
            print("-", doc_id)

    print("\n" + "=" * 65)
    print("INTERPRETATION")
    print("=" * 65)

    if not common_ids and not common_questions and not common_pairs:
        print("No direct question-level leakage detected.")
    else:
        print("Potential question-level overlap detected.")
        print("Review overlapping records before deciding whether to remove them.")

    if common_documents:
        print(
            "Some documents occur in both splits. "
            "This is not automatically leakage because multiple "
            "questions may legitimately refer to the same document."
        )
    else:
        print("No shared document IDs detected.")

    print("\nSTEP 11 ANALYSIS COMPLETED")


check_split_leakage(train_df, dev_df)
```

```python

import pandas as pd
import re

def normalize_text(value):
    if not isinstance(value, str):
        return ""
    return re.sub(r"\s+", " ", value).strip().lower()

# Work on copies
train_check = train_df.copy()
dev_check = dev_df.copy()

# Normalize questions and document IDs
for df in (train_check, dev_check):
    df["NORMALIZED_QUESTION"] = df["QUESTION_TEXT"].apply(normalize_text)
    df["NORMALIZED_DOCUMENT"] = df["DOCUMENT"].apply(
        lambda x: str(x).strip() if pd.notna(x) else ""
    )

# --------------------------------------------------
# 1. QUESTION OVERLAP
# --------------------------------------------------

common_questions = (
    set(train_check["NORMALIZED_QUESTION"])
    & set(dev_check["NORMALIZED_QUESTION"])
) - {""}

question_overlaps = pd.concat([
    train_check[
        train_check["NORMALIZED_QUESTION"].isin(common_questions)
    ].assign(SPLIT="Training"),

    dev_check[
        dev_check["NORMALIZED_QUESTION"].isin(common_questions)
    ].assign(SPLIT="Development")
], ignore_index=True)

print("COMMON QUESTIONS:", len(common_questions))

display(
    question_overlaps.sort_values(
        ["NORMALIZED_QUESTION", "SPLIT"]
    )[
        ["SPLIT", "QUESTION_ID", "QUESTION_TEXT",
         "DOCUMENT", "ANSWER", "ANSWERABLE"]
    ]
)

# --------------------------------------------------
# 2. QUESTION-DOCUMENT PAIR OVERLAP
# --------------------------------------------------

train_pairs = set(zip(
    train_check["NORMALIZED_QUESTION"],
    train_check["NORMALIZED_DOCUMENT"]
))

dev_pairs = set(zip(
    dev_check["NORMALIZED_QUESTION"],
    dev_check["NORMALIZED_DOCUMENT"]
))

common_pairs = train_pairs & dev_pairs

train_pair_mask = [
    pair in common_pairs
    for pair in zip(
        train_check["NORMALIZED_QUESTION"],
        train_check["NORMALIZED_DOCUMENT"]
    )
]

dev_pair_mask = [
    pair in common_pairs
    for pair in zip(
        dev_check["NORMALIZED_QUESTION"],
        dev_check["NORMALIZED_DOCUMENT"]
    )
]

pair_overlaps = pd.concat([
    train_check[train_pair_mask].assign(SPLIT="Training"),
    dev_check[dev_pair_mask].assign(SPLIT="Development")
], ignore_index=True)

print("\nCOMMON QUESTION-DOCUMENT PAIRS:", len(common_pairs))

display(
    pair_overlaps.sort_values(
        ["NORMALIZED_QUESTION", "SPLIT"]
    )[
        ["SPLIT", "QUESTION_ID", "QUESTION_TEXT",
         "DOCUMENT", "ANSWER", "ANSWERABLE"]
    ]
)

print("\nSTEP 11 OVERLAP INVESTIGATION COMPLETED")

```

```python
# STEP 11: OVERLAP CONFLICT SUMMARY

question_summary = (
    question_overlaps
    .groupby("NORMALIZED_QUESTION")
    .agg(
        split_count=("SPLIT", "nunique"),
        question_ids=("QUESTION_ID", lambda x: list(x)),
        documents=("NORMALIZED_DOCUMENT", lambda x: list(x)),
        answerability=("ANSWERABLE", lambda x: list(x)),
        answers=("ANSWER", lambda x: list(x))
    )
    .reset_index()
)

question_summary["DOCUMENT_CONFLICT"] = question_summary["documents"].apply(
    lambda x: len(set(x)) > 1
)

question_summary["ANSWERABILITY_CONFLICT"] = question_summary["answerability"].apply(
    lambda x: len(set(map(str, x))) > 1
)

question_summary["ANSWER_CONFLICT"] = question_summary["answers"].apply(
    lambda x: len(set(str(a).strip() for a in x)) > 1
)

print("OVERLAP CLASSIFICATION")
print("Repeated questions:", len(question_summary))
print("Different documents:", question_summary["DOCUMENT_CONFLICT"].sum())
print("Different answerability:", question_summary["ANSWERABILITY_CONFLICT"].sum())
print("Different answers:", question_summary["ANSWER_CONFLICT"].sum())

display(question_summary[
    question_summary[
        ["DOCUMENT_CONFLICT", "ANSWERABILITY_CONFLICT", "ANSWER_CONFLICT"]
    ].any(axis=1)
])
```

# 12 – Reproducible few-shot subsets

```python
import os
import pandas as pd

RANDOM_SEED = 42

# Preserve the original datasets
train_original = train_df.copy()
dev_original = dev_df.copy()

# Create reproducible, stratified samples
answerable = train_df[train_df["ANSWERABLE"] == "Y"].sample(
    frac=1, random_state=RANDOM_SEED
)

unanswerable = train_df[train_df["ANSWERABLE"] == "N"].sample(
    frac=1, random_state=RANDOM_SEED
)

# Nested subsets with proportional class representation
few_shot_subsets = {
    "1_percent": pd.concat([answerable.iloc[:5], unanswerable.iloc[:1]]),
    "5_percent": pd.concat([answerable.iloc[:23], unanswerable.iloc[:7]]),
    "10_percent": pd.concat([answerable.iloc[:45], unanswerable.iloc[:15]])
}

# Shuffle each subset deterministically
for name in few_shot_subsets:
    few_shot_subsets[name] = few_shot_subsets[name].sample(
        frac=1, random_state=RANDOM_SEED
    ).reset_index(drop=True)

# Save processed subsets
output_dir = "../../data/processed/TechQA"
os.makedirs(output_dir, exist_ok=True)

for name, subset in few_shot_subsets.items():
    subset.to_parquet(
        os.path.join(output_dir, f"train_{name}.parquet"),
        index=False
    )

    print(f"{name}: {len(subset)} records")
    print(subset["ANSWERABLE"].value_counts().to_dict())

print("\nSTEP 12 — FEW-SHOT SUBSETS CREATED")
```

```python

# STEP 14 — TECHQA DOCUMENT CORPUS PARSING AND VALIDATION

import os
import re
import pandas as pd
from pathlib import Path

# --------------------------------------------------
# 1. Locate corpus automatically
# --------------------------------------------------

current = Path.cwd().resolve()
corpus_path = None

for folder in [current, *current.parents]:
    candidate = (
        folder / "data" / "raw" / "TechQA" /
        "technote_corpus" / "technotes.id_title_text"
    )
    if candidate.is_file():
        corpus_path = candidate
        break

if corpus_path is None:
    raise FileNotFoundError("TechQA document corpus not found.")

print("Corpus:", corpus_path)
print("Corpus size (GB):", round(corpus_path.stat().st_size / (1024**3), 2))

# --------------------------------------------------
# 2. Collect required document IDs
# --------------------------------------------------

required_ids = set()

for df in [train_df, dev_df]:
    if "DOCUMENT" not in df.columns:
        raise KeyError("DOCUMENT column missing from dataset.")

    ids = df["DOCUMENT"].dropna().astype(str).str.strip()
    required_ids.update(x for x in ids if x not in {"", "-"})

print("\nUnique required document IDs:", len(required_ids))

# --------------------------------------------------
# 3. Parse labelled corpus sections
# --------------------------------------------------

header_pattern = re.compile(
    r"^\s*=+\s*(ID|TITLE|TEXT)\s*=+\s*$",
    re.IGNORECASE
)

documents = {}
total_records = 0

record = {}
current_field = None
text_parts = []

def save_record():
    global total_records, record, text_parts

    doc_id = str(record.get("DOC_ID", "")).strip()

    if doc_id:
        total_records += 1

        if doc_id in required_ids:
            documents[doc_id] = {
                "DOC_ID": doc_id,
                "TITLE": record.get("TITLE", "").strip(),
                "TEXT": "".join(text_parts).strip()
            }

with open(corpus_path, "r", encoding="utf-8", errors="replace") as f:

    for line in f:
        match = header_pattern.match(line.rstrip("\r\n"))

        if match:
            field = match.group(1).upper()

            if field == "ID":
                if record:
                    save_record()

                record = {}
                text_parts = []

            current_field = field
            continue

        if current_field == "ID":
            if "DOC_ID" not in record:
                record["DOC_ID"] = line.strip()

        elif current_field == "TITLE":
            if "TITLE" not in record:
                record["TITLE"] = line.strip()

        elif current_field == "TEXT":
            text_parts.append(line)

    if record:
        save_record()

# --------------------------------------------------
# 4. Create matched document dataframe
# --------------------------------------------------

documents_df = pd.DataFrame(
    list(documents.values()),
    columns=["DOC_ID", "TITLE", "TEXT"]
)

found_ids = set(documents.keys())
missing_ids = required_ids - found_ids

# --------------------------------------------------
# 5. Validate training and development references
# --------------------------------------------------

def validate_documents(df, split_name):
    references = (
        df.loc[df["ANSWERABLE"].astype(str).str.upper().eq("Y"), "DOCUMENT"]
        .dropna()
        .astype(str)
        .str.strip()
    )

    references = references[~references.isin(["", "-"])]

    found = references.isin(found_ids)

    report = pd.DataFrame({
        "DOCUMENT": references.values,
        "FOUND": found.values
    })

    print(f"\n{split_name} DOCUMENT VALIDATION")
    print("Answerable references:", len(references))
    print("Documents found:", int(found.sum()))
    print("Documents not found:", int((~found).sum()))

    return report

train_document_report = validate_documents(train_df, "TRAIN")
dev_document_report = validate_documents(dev_df, "DEVELOPMENT")

# --------------------------------------------------
# 6. Save validation outputs
# --------------------------------------------------

output_dir = Path("data/processed/TechQA")
output_dir.mkdir(parents=True, exist_ok=True)

documents_df.to_parquet(
    output_dir / "matched_documents.parquet",
    index=False
)

train_document_report.to_csv(
    output_dir / "train_document_report.csv",
    index=False
)

dev_document_report.to_csv(
    output_dir / "dev_document_report.csv",
    index=False
)

pd.DataFrame({
    "metric": [
        "Total corpus records parsed",
        "Unique required document IDs",
        "Documents matched",
        "Documents missing"
    ],
    "value": [
        total_records,
        len(required_ids),
        len(found_ids),
        len(missing_ids)
    ]
}).to_json(
    output_dir / "document_validation_metadata.json",
    orient="records",
    indent=4
)

# --------------------------------------------------
# 7. Final results
# --------------------------------------------------

print("\nDOCUMENT CORPUS SUMMARY")
print("Total corpus records parsed:", total_records)
print("Required document IDs:", len(required_ids))
print("Matched documents:", len(found_ids))
print("Missing documents:", len(missing_ids))

print("\nSAMPLE MATCHED DOCUMENTS")
display(documents_df.head())

if missing_ids:
    print("\nSAMPLE MISSING DOCUMENT IDS:")
    print(list(missing_ids)[:10])

print("\nSAVED FILES:")
for file in sorted(output_dir.iterdir()):
    if file.is_file():
        print(file.name, "|", round(file.stat().st_size / 1024, 2), "KB")

print("\nSTEP 14 DOCUMENT CORPUS PARSING AND VALIDATION COMPLETED")
```

# 15 — Final Preprocessing Integration and Dataset Handoff

```python

import pandas as pd
import json
from pathlib import Path

# ============================================================
# STEP 15: FINAL PREPROCESSING INTEGRATION AND DATASET HANDOFF
# ============================================================

ROOT = Path.cwd().parent.parent

DATA_DIR = ROOT / "data" / "processed" / "TechQA"
DOC_DIR = ROOT / "notebooks" / "02_data_preprocessing" / "data" / "processed" / "TechQA"

train_path = DATA_DIR / "train_processed.parquet"
dev_path = DATA_DIR / "dev_processed.parquet"
documents_path = DOC_DIR / "matched_documents.parquet"

# 1. Verify and load files
for path in [train_path, dev_path, documents_path]:
    if not path.exists():
        raise FileNotFoundError(f"Required file not found: {path}")

train = pd.read_parquet(train_path)
dev = pd.read_parquet(dev_path)
documents = pd.read_parquet(documents_path)

print("FILES LOADED")
print("Train:", train.shape)
print("Development:", dev.shape)
print("Matched documents:", documents.shape)

# 2. Validate required columns
required_question_cols = [
    "DOCUMENT", "ANSWER", "ANSWERABLE",
    "START_OFFSET", "END_OFFSET"
]

for name, df in [("Train", train), ("Development", dev)]:
    missing = [c for c in required_question_cols if c not in df.columns]
    if missing:
        raise ValueError(f"{name} missing columns: {missing}")

for col in ["DOC_ID", "TEXT"]:
    if col not in documents.columns:
        raise ValueError(f"Matched documents missing column: {col}")

# 3. Create document lookup
documents["DOC_ID"] = documents["DOC_ID"].astype(str).str.strip()

document_lookup = (
    documents.drop_duplicates("DOC_ID")
    .set_index("DOC_ID")["TEXT"]
)

# 4. Attach document text
def attach_document_text(df):
    result = df.copy()
    result["DOCUMENT"] = result["DOCUMENT"].astype(str).str.strip()
    result["DOCUMENT_TEXT"] = result["DOCUMENT"].map(document_lookup)
    return result

train_final = attach_document_text(train)
dev_final = attach_document_text(dev)

# 5. Document matching validation
def document_report(df, split):
    answerable = (
        df["ANSWERABLE"].astype(str).str.strip().str.upper()
        .isin(["Y", "YES", "TRUE", "1"])
    )

    found = (
        df["DOCUMENT_TEXT"].notna()
        & df["DOCUMENT_TEXT"].astype(str).str.strip().ne("")
    )

    report = {
        "split": split,
        "total_records": len(df),
        "answerable_records": int(answerable.sum()),
        "documents_found": int((answerable & found).sum()),
        "documents_missing": int((answerable & ~found).sum()),
        "unanswerable_records": int((~answerable).sum())
    }

    print(f"\n{split.upper()} DOCUMENT VALIDATION")
    for key, value in report.items():
        if key != "split":
            print(f"{key}: {value}")

    return report

train_report = document_report(train_final, "train")
dev_report = document_report(dev_final, "development")

# 6. Answer offset validation
def validate_offsets(df, split):
    statuses = []

    for _, row in df.iterrows():
        answer = row["ANSWER"]
        text = row["DOCUMENT_TEXT"]
        start = row["START_OFFSET"]
        end = row["END_OFFSET"]

        if pd.isna(text) or not str(text).strip():
            statuses.append("Document missing")
            continue

        if pd.isna(answer) or not str(answer).strip():
            statuses.append("Answer missing")
            continue

        try:
            start, end = int(start), int(end)
        except (ValueError, TypeError):
            statuses.append("Invalid offsets")
            continue

        text = str(text)
        answer = str(answer)

        if start < 0 or end > len(text) or start >= end:
            statuses.append("Out of range")
        elif text[start:end] == answer:
            statuses.append("Valid")
        else:
            statuses.append("Text mismatch")

    report = pd.Series(statuses).value_counts()

    print(f"\n{split.upper()} OFFSET VALIDATION")
    print(report.to_string())

    return {str(k): int(v) for k, v in report.items()}

train_offset_report = validate_offsets(train_final, "train")
dev_offset_report = validate_offsets(dev_final, "development")

# 7. Save final integrated datasets
train_final.to_parquet(DATA_DIR / "train_final.parquet", index=False)
dev_final.to_parquet(DATA_DIR / "dev_final.parquet", index=False)

# 8. Save metadata
metadata = {
    "step": 15,
    "description": "Final TechQA preprocessing integration and handoff",
    "train": train_report,
    "development": dev_report,
    "matched_document_count": int(len(documents)),
    "train_offset_validation": train_offset_report,
    "development_offset_validation": dev_offset_report
}

with open(DATA_DIR / "step15_metadata.json", "w") as f:
    json.dump(metadata, f, indent=4)

# 9. Final output summary
print("\nSAVED FILES")

for filename in [
    "train_final.parquet",
    "dev_final.parquet",
    "step15_metadata.json"
]:
    path = DATA_DIR / filename
    print(f"{filename} | {path.stat().st_size / 1024:.2f} KB")

print("\nSTEP 15 EXECUTION COMPLETED")
```
